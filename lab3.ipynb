```python
age = int(input("Enter your age: "))

if age < 13:
    print("Child") # [cite: 10]
elif 13 <= age <= 19:
    print("Teen") # [cite: 11]
elif 20 <= age <= 59:
    print("Adult") # [cite: 14]
else:
    print("Senior") # [cite: 13]
```

```python
num = float(input("Enter a number: "))

# التحقق من كونه موجباً
if num > 0:
    print("Positive") # [cite: 21]
else:
    print("Not Positive") # [cite: 22]

# التحقق من النطاق (10-50)
if 10 <= num <= 50:
    print("The number is between 10 and 50 (inclusive).") # [cite: 23]
```

```python
color = input("Enter a colour name (red, yellow, green): ").lower()

if color != "": # التأكد أن الإدخال ليس فارغاً [cite: 27]
    match color: # [cite: 26]
        case "red":
            print("Stop")
        case "yellow":
            print("Slow down")
        case "green":
            print("Go")
        case _:
            print("Invalid color")
else:
    print("Input cannot be empty")
```

```python
print("Menu:")
print("1. Addition") # [cite: 32]
print("2. Subtraction") # [cite: 33]
print("3. Multiplication") # [cite: 34]
print("4. Division") # [cite: 35]

choice = input("Choose an option (1-4): ")
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

match choice: # [cite: 37]
    case "1":
        print(f"Result: {num1 + num2}")
    case "2":
        print(f"Result: {num1 - num2}")
    case "3":
        print(f"Result: {num1 * num2}")
    case "4":
        if num2 != 0:
            print(f"Result: {num1 / num2}")
        else:
            print("Cannot divide by zero")
```

```python
score = float(input("Enter student score: "))

# تحديد النجاح أو الرسوب [cite: 41]
if score >= 60:
    print("Status: Passed")
else:
    print("Status: Failed")

# تعيين الدرجة [cite: 41]
if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"

# رسالة بناءً على الدرجة باستخدام match [cite: 42]
match grade:
    case "A":
        print("Excellent work!")
    case "B":
        print("Good job!")
    case "C":
        print("You passed.")
    case "F":
        print("Better luck next time.")
```
