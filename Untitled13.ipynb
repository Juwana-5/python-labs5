```python
# طلب 5 أرقام من المستخدم وتخزينها في قائمة
numbers = []
for i in range(5):
    num = float(input(f"Enter number {i+1}: "))
    numbers.append(num)

# حساب المجموع، المتوسط، القيمة القصوى والدنيا
total_sum = sum(numbers)
average = total_sum / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

# طباعة النتائج
print(f"List: {numbers}")
print(f"Sum: {total_sum}")
print(f"Average: {average}")
print(f"Maximum: {maximum}")
print(f"Minimum: {minimum}")
```

```python
# طلب 5 كلمات وتخزينها في Tuple
words_list = []
for i in range(5):
    word = input(f"Enter word {i+1}: ")
    words_list.append(word)

words_tuple = tuple(words_list)

# طباعة الكلمة الأولى، الأخيرة، والعدد الإجمالي
print(f"First word: {words_tuple[0]}")
print(f"Last word: {words_tuple[-1]}")
print(f"Total number of words: {len(words_tuple)}")

# التحقق من وجود كلمة معينة
search_word = input("Enter a word to check if it's in the tuple: ")
if search_word in words_tuple:
    print(f"Yes, '{search_word}' is in the tuple.")
else:
    print(f"No, '{search_word}' is not in the tuple.")
```

```python
# إنشاء قاموس وتعبئته ببيانات 3 طلاب
students_scores = {}

for i in range(3):
    name = input(f"Enter name of student {i+1}: ")
    score = float(input(f"Enter score for {name}: "))
    students_scores[name] = score

# طباعة جميع الطلاب ودرجاتهم
print("\nStudents and their scores:")
for name, score in students_scores.items():
    print(f"{name}: {score}")

# إيجاد الطالب صاحب أعلى درجة
top_student = max(students_scores, key=students_scores.get)
print(f"\nStudent with the highest score: {top_student} ({students_scores[top_student]})")
```
