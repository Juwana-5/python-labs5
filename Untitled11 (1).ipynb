```python
import math

# طلب نصف القطر من المستخدم [cite: 4]
radius = float(input("Enter the radius of the circle: "))

# دالة لحساب المساحة [cite: 5]
def calculate_area(r):
    return math.pi * pow(r, 2)

# lambda لحساب المحيط [cite: 6]
calculate_circumference = lambda r: 2 * math.pi * r

area = calculate_area(radius)
circumference = calculate_circumference(radius)

# طباعة النتائج بـ 2 خانة عشرية [cite: 8]
print(f"Area: {round(area, 2)}")
print(f"Circumference: {round(circumference, 2)}")
```

```python
import re

# طلب نص من المستخدم [cite: 10]
text = input("Enter a string containing numbers: ")

# استخدام Regex لإيجاد كل الأرقام [cite: 11]
numbers = re.findall(r'\d+', text)
# تحويل الأرقام من نصوص إلى أرقام صحيحة
numbers_int = [int(n) for n in numbers]

# دالة لجمع الأرقام [cite: 12]
def sum_numbers(nums_list):
    return sum(nums_list)

# lambda لمضاعفة المجموع [cite: 13]
double_sum = lambda s: s * 2

total = sum_numbers(numbers_int)
doubled = double_sum(total)

# طباعة النتائج [cite: 14]
print(f"Numbers found: {numbers_int}")
print(f"Sum: {total}")
print(f"Doubled value: {doubled}")
```

```python
import math

# طلب أرقام مفصولة بفاصلة [cite: 16]
user_input = input("Enter numbers separated by commas: ")
# تحويل المدخلات إلى قائمة floats [cite: 17]
numbers = [float(x.strip()) for x in user_input.split(',')]

# دالة لتربيع الأرقام باستخدام lambda [cite: 18]
squares = list(map(lambda x: x**2, numbers))

# استخدام math لحساب النتائج [cite: 19]
max_square = max(squares) # [cite: 20]
sqrt_of_sum = math.sqrt(sum(squares)) # [cite: 21]

# طباعة النتائج [cite: 22]
print(f"Squares: {squares}")
print(f"Maximum square: {max_square}")
print(f"Square root of the sum of squares: {round(sqrt_of_sum, 2)}")
```

```python
import re
import math

# التحقق من كلمة المرور [cite: 24]
password = input("Enter your password: ")
has_upper = re.search(r'[A-Z]', password) # [cite: 26]
has_lower = re.search(r'[a-z]', password) # [cite: 27]
has_num = re.search(r'\d', password) # [cite: 28]

if has_upper and has_lower and has_num:
    strength = "Strong"
else:
    strength = "Weak"

# حساب متوسط 5 درجات [cite: 29]
scores_input = input("Enter 5 scores separated by spaces: ")
scores = [float(s) for s in scores_input.split()]

# دالة و lambda لحساب المتوسط [cite: 30]
calc_avg = lambda s_list: sum(s_list) / len(s_list)
average = calc_avg(scores)

# تقريب المتوسط باستخدام math.ceil [cite: 31]
rounded_avg = math.ceil(average)

# الطباعة النهائية [cite: 32]
print(f"Password status: {strength}")
print(f"Rounded average score: {rounded_avg}")
```

```python
import re
import math

# طلب فقرة من المستخدم [cite: 34]
paragraph = input("Enter a paragraph: ")

# إيجاد الكلمات التي تبدأ بحرف علة [cite: 35]
vowel_words = re.findall(r'\b[aeiouAEIOU]\w*', paragraph)

# دالة لحساب إجمالي حروف العلة في الفقرة [cite: 36]
def count_vowels(text):
    return len(re.findall(r'[aeiouAEIOU]', text))

# lambda لحساب طول كل كلمة تبدأ بحرف علة [cite: 37]
word_lengths = list(map(lambda w: len(w), vowel_words))

# حساب متوسط طول الكلمات [cite: 38]
if word_lengths:
    avg_len = sum(word_lengths) / len(word_lengths)
else:
    avg_len = 0

# الطباعة [cite: 39]
print(f"Words starting with vowels: {vowel_words}")
print(f"Total vowels in paragraph: {count_vowels(paragraph)}")
print(f"Average vowel-word length: {round(avg_len, 2)}")
```
