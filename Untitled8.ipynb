```python
total = 0
for i in range(10): # طلب 10 أرقام
    num = int(input("أدخل رقماً: "))
    
    if num == 0: # التوقف إذا كان الرقم 0
        break
    if num < 0: # تخطي الأرقام السالبة
        continue
    if num % 2 == 0: # التحقق إذا كان زوجياً
        total += num # استخدام += (الجمع التراكمي)

print("المجموع النهائي هو:", total) # طباعة المجموع
```

```python
number = int(input("أدخل رقماً لجدول الضرب: ")) # [cite: 10]

for i in range(1, 13): # من 1 إلى 12 [cite: 11]
    product = number * i
    
    if product > 50: # التوقف إذا كان الناتج أكبر من 50 [cite: 13]
        break
    if i % 3 == 0: # تخطي مضاعفات الـ 3 [cite: 12]
        continue
        
    print(f"{number} x {i} = {product}")
```

```python
secret_number = 7 # اختيار رقم سري بين 1 و 20 [cite: 16]
attempts = 0

while True: # حلقة غير منتهية حتى التخمين الصحيح [cite: 18]
    guess = int(input("خمن الرقم السري (بين 1 و 20): ")) # [cite: 17]
    
    if guess < 1 or guess > 20: # تخطي الأرقام خارج النطاق [cite: 20]
        print("رقم غير صالح!")
        continue
    
    attempts += 1 # زيادة عدد المحاولات الصحيحة [cite: 21]
    
    if guess == secret_number: # التوقف عند التخمين الصحيح [cite: 19]
        print(f"مبروك! خمنت الرقم في {attempts} محاولات.")
        break
```

```python
n = int(input("أدخل قيمة n: ")) # [cite: 24]
count = 0

for i in range(1, n + 1):
    if i > 50: # التوقف إذا وجد رقم أكبر من 50 [cite: 26]
        break
    if i % 8 == 0: # تخطي الأرقام التي تقبل القسمة على 8 [cite: 27]
        continue
    if i % 4 == 0: # عد الأرقام التي تقبل القسمة على 4 [cite: 25]
        count += 1

print("العدد الإجمالي هو:", count) # [cite: 28]
```

```python
for i in range(1, 6): # الصفوف من 1 إلى 5 [cite: 31]
    for j in range(1, i + 1): # الأعمدة من 1 إلى i [cite: 31]
        if j == 3: # تخطي طباعة الرقم 3 [cite: 32]
            continue
        if i * j > 6: # التوقف إذا كان حاصل الضرب أكبر من 6 [cite: 33]
            break
        print(j, end=" ")
    print() # سطر جديد لكل صف
```
