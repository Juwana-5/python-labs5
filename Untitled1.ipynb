```python
import os

# 1. إنشاء مجلد lab_data والتحقق من وجوده
parent = "lab_data"
if not os.path.exists(parent):
    os.makedirs(parent)
    print(f"Directory '{parent}' created.")

# 2. إنشاء المجلدات الفرعية (input, output, logs)
for sub in ["input", "output", "logs"]:
    os.makedirs(os.path.join(parent, sub), exist_ok=True)

# 3. عرض المحتويات وتحديد النوع
print("\nItems in lab_data:")
for item in os.listdir(parent):
    path = os.path.join(parent, item)
    kind = "Directory" if os.path.isdir(path) else "File"
    print(f"- {item}: {kind}")

# 4. إعادة تسمية logs إلى history
os.rename(os.path.join(parent, "logs"), os.path.join(parent, "history"))
print("\nRenamed 'logs' to 'history'.")
```

```python
# 1. إنشاء ملف notes.txt وكتابة 5 أسطر
input_path = os.path.join("lab_data", "input", "notes.txt")
with open(input_path, "w") as f:
    f.write("Line 1: Python Lab 8\nLine 2: Operating Systems\nLine 3: File Handling\nLine 4: Subdirectories\nLine 5: Automation\n")

# 2. قراءة الملف وطباعة عدد الأسطر
with open(input_path, "r") as f:
    lines = f.readlines()
    print(f"Total lines in notes.txt: {len(lines)}")

# 3. نسخ المحتوى إلى summary.txt
output_path = os.path.join("lab_data", "output", "summary.txt")
with open(output_path, "w") as f:
    f.writelines(lines)
```

```python
import pyfiglet

# 1. إنشاء فن ASCII (اكتبي اسمك بدلاً من Jwana)
result = pyfiglet.figlet_format("Jwana")
print(result)

# 2. حفظ النتيجة في ملف ascii_art.txt
with open(os.path.join("lab_data", "output", "ascii_art.txt"), "w") as f:
    f.write(result)
```
